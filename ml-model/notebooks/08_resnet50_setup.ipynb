```python
model = resnet50_model.build_resnet50_model(
    trainable=False
)

print("ResNet50 model created successfully!")
```

```python
print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)
```

```python
model.summary()
```

```python
base_model = model.get_layer("resnet50")

print("ResNet50 layers:", len(base_model.layers))

trainable_count = sum(
    layer.trainable
    for layer in base_model.layers
)

frozen_count = sum(
    not layer.trainable
    for layer in base_model.layers
)

print("Frozen layers:", frozen_count)
print("Trainable layers:", trainable_count)
```

```python
model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-3
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

print("ResNet50 compiled successfully!")
```
